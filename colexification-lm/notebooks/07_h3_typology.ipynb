**Pre-specified design (D-09 amendment, D-25, D-26, D-56–D-58).** Statistic: per-family pair-level AUC (within-family colexified vs non-colexified pairs among concepts the family attests). Inference: joint concept-label permutation (10,000), concept-bootstrap CI, BH across conditions × families. Confound control: rarefaction to 20 languages × 200. Confirmatory test: Spearman across families between rarefied AUC and the Wikipedia proxy, one-sided, BH across conditions.
