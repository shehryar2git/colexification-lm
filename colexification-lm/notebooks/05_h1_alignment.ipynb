# 05 · H1: does model space mirror typological space?

Narrative only: the logic lives in `src/analysis/h1_alignment.py` and `src/utils/stats.py`. Run `python run.py analyse` first.

**Pre-specified (confirmatory):** one-sided Mantel test of Spearman ρ between typological distance (weighted shortest path) and model distance (1 − centred cosine), 10,000 joint concept permutations, Benjamini–Hochberg across conditions. **H1 is supported for a condition** if the BH-adjusted p < .05 and ρ > 0. Everything else here is labelled exploratory or descriptive.
